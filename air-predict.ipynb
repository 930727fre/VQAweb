```python
# Takes the path to an RGB image file as input, reads the image using Keras library and converts it to a NumPy array. 

def build_x(path, y, df):
    train_img = []
    for i in range(len(y)):
        img = image.load_img(path + df['Filename'][i])
        img = image.img_to_array(img)
        img = img / 255
        train_img.append(img)

    x = np.array(train_img)
    return x
```

```python
#from tensorflow.keras.models import Model
#from tensorflow.keras.layers import Input, Conv2D, LeakyReLU, MaxPooling2D, Flatten, Dense, Add

# Define input
inputs = Input(shape=(224, 224, 3))

# Convolution Layer
conv1 = Conv2D(64, (3, 3), padding='same', name='block1_conv1')(inputs)
leak1 = LeakyReLU(alpha=0.1)(conv1)
conv2 = Conv2D(64, (3, 3), padding='same', name='block1_conv2')(leak1)
leak2 = LeakyReLU(alpha=0.1)(conv2)
pool1 = MaxPooling2D((3, 3), strides=(2, 2), name='block1_pool')(leak2)

conv3 = Conv2D(128, (3, 3), padding='same', name='block2_conv1')(pool1)
leak3 = LeakyReLU(alpha=0.1)(conv3)
conv4 = Conv2D(128, (3, 3), padding='same', name='block2_conv2')(leak3)
leak4 = LeakyReLU(alpha=0.1)(conv4)
pool2 = MaxPooling2D((3, 3), strides=(2, 2), name='block2_pool')(leak4)

conv5 = Conv2D(128, (3, 3), padding='same', name='block3_conv1')(pool2)
leak5 = LeakyReLU(alpha=0.1)(conv5)
res2 = Add()([leak5, pool2])
pool3 = MaxPooling2D((3, 3), strides=(2, 2), name='block3_pool')(res2)

conv7 = Conv2D(128, (3, 3), padding='same', name='block4_conv1')(pool3)
leak7 = LeakyReLU(alpha=0.1)(conv7)
res3 = Add()([leak7, pool3])
pool4 = MaxPooling2D((3, 3), strides=(2, 2), name='block4_pool')(res3)

conv9 = Conv2D(128, (3, 3), padding='same', name='block5_conv1')(pool4)
leak9 = LeakyReLU(alpha=0.1)(conv9)
res4 = Add()([leak9, pool4])
pool5 = MaxPooling2D((3, 3), strides=(2, 2), name='block5_pool')(res4)

conv11 = Conv2D(256, (3, 3), padding='same', name='block6_conv1')(pool5)
leak11 = LeakyReLU(alpha=0.1)(conv11)
conv12 = Conv2D(256, (3, 3), padding='same', name='block6_conv2')(leak11)
leak12 = LeakyReLU(alpha=0.1)(conv12)
pool6 = MaxPooling2D((3, 3), strides=(2, 2), name='block6_pool')(leak12)

# Flatten Layer
flatten = Flatten()(pool6)

# Fully Connected Layer
dense1 = Dense(1024, activation='relu')(flatten)
dense2 = Dense(1024, activation='relu')(dense1)

# Output Layer
pm25_output = Dense(1, activation='linear', name='PM2.5_output')(dense2)
pm10_output = Dense(1, activation='linear', name='PM10_output')(dense2)
o3_output = Dense(1, activation='linear', name='O3_output')(dense2)
co_output = Dense(1, activation='linear', name='CO_output')(dense2)
so2_output = Dense(1, activation='linear', name='SO2_output')(dense2)
no2_output = Dense(1, activation='linear', name='NO2_output')(dense2)

# Define model
model = Model(inputs=inputs, outputs=[pm25_output, pm10_output, o3_output, co_output, so2_output, no2_output])

optimizer = Adam(learning_rate=1e-4, clipnorm=1.0)  # 使用clipnorm或clipvalue限制梯度大小
#model.compile(optimizer=optimizer, loss='mae')
# Compile model
model.compile(optimizer=optimizer, loss='mae')
model.summary()
```

```python
#from tensorflow.keras.callbacks import ModelCheckpoint
from glob import glob
from tensorflow.keras.callbacks import ReduceLROnPlateau

# Weight
saved_weights_dir = "/kaggle/input/air-weight-41/"
saved_weights = sorted(glob(os.path.join(saved_weights_dir, "model_weights_epoch_*.weights.h5")))

if saved_weights:
    latest_weights = saved_weights[-1]
    print(f"加載權重：{latest_weights}")
    model.load_weights(latest_weights)
    initial_epoch = int(latest_weights.split('_')[-1].split('.')[0])  # 獲取初始epoch
else:
    print("沒有找到已保存的權重，從頭開始訓練")
    initial_epoch = 0

# Path of weight
weight_path = "model_weights_epoch_{epoch:03d}.weights.h5"

# Callback Function
callbacks = [
    ModelCheckpoint(filepath=weight_path, 
                    monitor='val_loss',        
                    save_weights_only=True,    # Store weight
                    save_freq='epoch',         # Each epoch
                    verbose=1,                 # Show progress
                    save_best_only=False)      # Store each file
]

lr_callback = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, verbose=1)
callbacks.append(lr_callback)
```

```python
history = model.fit(
    train_img, 
    [y_train['PM2.5'], y_train['PM10'], y_train['O3'], y_train['CO'], y_train['SO2'], y_train['NO2']],
    validation_data=(val_img, [y_val['PM2.5'], y_val['PM10'], y_val['O3'], y_val['CO'], y_val['SO2'], y_val['NO2']]),
    batch_size=16, 
    epochs=100, 
    initial_epoch=initial_epoch,
    callbacks=callbacks
)

```
